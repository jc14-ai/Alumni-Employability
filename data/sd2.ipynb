```python
df = df.drop(columns=['Student Number'])
df = df.drop(columns=df.columns[df.isna().sum() == len(df)])
df.head()
```

```python
df.isna().sum()
```

```python
df.isna().mean() * 100
```

```python
df.groupby("Degree").size()
```

```python
df = df.fillna(0)
df.isna().sum()
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
encoded_degree = encoder.fit_transform(df[["Degree"]])
encoded_gender = encoder.fit_transform(df[["Gender"]])
encoded_leadership_pos = encoder.fit_transform(df[["Leadership POS"]])
encoded_act_member_pos = encoder.fit_transform(df[["Act Member POS"]])

df['Degree'] = encoded_degree[:]
df['Gender'] = encoded_gender[:]
df['Leadership POS'] = encoded_leadership_pos[:]
df['Act Member POS'] = encoded_act_member_pos[:]
df.head(n=20)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(df)
df['Employability'] = kmeans.labels_
df.head()
```

```python
df['Employability'] = df['Employability'].map({0: 'Not Employable', 1: 'Employable'})
```
